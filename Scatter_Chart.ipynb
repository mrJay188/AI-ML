```python
import random
from matplotlib import style
days=[]
pune_temp=[]
rajkot_temp=[]
for i in range(16):
    days.append(i)

for _ in range(16):
    pune_temp.append(random.randint(0, 30))
    rajkot_temp.append(random.randint(0, 30))
display(days)
display(pune_temp)
display(rajkot_temp)
```

```python
m1_stu_age = np.random.randint(16,30,(100))
py_stu_age = np.random.randint(14,28,(100))

m1_stu_age
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(20, 10))


plt.subplot(3, 2, 1)

plt.plot(days, pune_temp,
         marker='o',
         linewidth=2,
         color='royalblue',
         label='Pune')

plt.plot(days, rajkot_temp,
         marker='o',
         linewidth=2,
         color='tomato',
         label='Rajkot')

plt.title("Temperature Analysis", fontsize=15, weight='bold')
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.grid(alpha=0.3)
plt.legend()



plt.subplot(3, 2, 2)

bins = np.arange(14, 31)

plt.hist(m1_stu_age,
         bins=bins,
         histtype='step',
         linewidth=3,
         color='#0077B6',
         label='Machine Learning')

plt.hist(py_stu_age,
         bins=bins,
         histtype='step',
         linewidth=3,
         color='#E63946',
         label='Python')

plt.title("Age Distribution", fontsize=15, weight='bold')
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.grid(alpha=0.3)
plt.legend(frameon=False)


plt.subplot(3, 2, 3)

courses = ['Python', 'ML', 'AI', 'DS']
students = [120, 95, 80, 105]

plt.bar(courses,
        students,
        color=['#4CAF50', '#2196F3', '#FF9800', '#9C27B0'])

plt.title("Students Enrolled", fontsize=15, weight='bold')
plt.ylabel("Students")
plt.grid(axis='y', alpha=0.3)


plt.subplot(3, 2, 4)

labels = ['Python', 'ML', 'AI', 'DS']
sizes = [35, 25, 20, 20]
colors = ['#4CAF50', '#2196F3', '#FF9800', '#9C27B0']

plt.pie(sizes,
        labels=labels,
        autopct='%1.1f%%',
        startangle=90,
        colors=colors)

plt.title("Course Popularity", fontsize=15, weight='bold')


plt.subplot(3, 2, 5)

plt.scatter(days,
            pune_temp,
            s=100,
            color='royalblue',
            label='Pune')

plt.scatter(days,
            rajkot_temp,
            s=100,
            color='tomato',
            label='Rajkot')

plt.title("Temperature Scatter", fontsize=15, weight='bold')
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.grid(alpha=0.3)
plt.legend()


plt.subplot(3, 2, 6)

plt.boxplot([m1_stu_age, py_stu_age],
            labels=['ML', 'Python'],
            patch_artist=True)

plt.title("Age Comparison", fontsize=15, weight='bold')
plt.ylabel("Age")
plt.grid(alpha=0.3)


plt.suptitle("Matplotlib Subplots Example", fontsize=20, weight='bold')

plt.tight_layout(rect=[0, 0, 1, 0.96])

plt.show()
```
