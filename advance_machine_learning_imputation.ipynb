```python
import pandas as pd
import numpy as np
data = {
    'Age': [25,np.nan,30,22,np.nan],
    'Score':[85,90,np.nan,75,80],
    'City':['New York','Paris','London',np.nan,'Paris']
}
dataf = pd.DataFrame(data)
dataf
```

```python
from sklearn.impute import SimpleImputer , KNNImputer

mean_imputer = SimpleImputer(strategy='mean')
dataf[['Age','Score']] = mean_imputer.fit_transform(dataf[['Age','Score']])
knn_imputer = KNNImputer(n_neighbors=2)
dataf_knn=knn_imputer.fit_transform(dataf[['Age','Score']])
```

```python
numerical_cols = df3_drop_row.select_dtypes(include=['int64','float64']).columns
num_plots = len(numerical_cols)
num_cols_per_row = 4
num_rows = (num_plots + num_cols_per_row - 1) // num_cols_per_row # Ceiling division

plt.figure(figsize=(num_cols_per_row * 4, num_rows * 3))
for i, var in enumerate(numerical_cols):
    plt.subplot(num_rows, num_cols_per_row, i + 1)
    sns.distplot(df3_drop_row[var])
    plt.title(f'Distribution of {var}')
plt.tight_layout()
plt.show()
```
