```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import PowerTransformer
```

```python
df = pd.DataFrame({'House_prices':[100000,120000,150000,20000,80000,250000]})
```

```python
df
```

```python
df['log_prices'] = np.log1p(df['House_prices'])
```

```python
df
```

```python
pt = PowerTransformer(method="yeo-johnson")
df['Prower_prices'] = pt.fit_transform(df[['House_prices']])
```
