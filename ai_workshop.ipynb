### `np.vstack()` - Vertical Stacking

`np.vstack()` stacks arrays in sequence vertically (row-wise). This means it concatenates arrays along the first axis (rows). The arrays must have the same number of columns (or the same shape along all but the first axis) to be vertically stacked.

```python
import numpy as np

arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([[7, 8, 9], [10, 11, 12]])

print("Array 1:")
print(arr1)
print("\nArray 2:")
print(arr2)

v_stack = np.vstack((arr1, arr2))
print("\nVertical Stack (vstack):")
print(v_stack)
```

### `np.hstack()` - Horizontal Stacking

`np.hstack()` stacks arrays in sequence horizontally (column-wise). This means it concatenates arrays along the second axis (columns). The arrays must have the same number of rows (or the same shape along all but the second axis) to be horizontally stacked.

```python
import numpy as np

arr3 = np.array([[1, 2,3,4,2], [3, 4,6,3,1]])
arr4 = np.array([[5, 6,6,7,5], [7, 8,14,5,6]])

print("Array 3:")
print(arr3)
print("\nArray 4:")
print(arr4)

h_stack = np.hstack((arr3, arr4))
print("\nHorizontal Stack (hstack):")
print(h_stack)
```

**Mathemetical Fundamentals**

```python
import numpy as np

# minimum and maximum
mn_arr = np.minimum(arr3, arr4)
mx_arr = np.maximum(arr3, arr4)

print("Minimum ")
print(mn_arr)

print("\n Maximum")
print(mx_arr)

print("\nSum of element-wise min and max arrays:")
print(mn_arr + mx_arr)

print("\nSubtraction of element-wise min and max arrays:")
print(mn_arr - mx_arr)

print("\nmultiplication of element-wise min and max arrays:")
print (mn_arr * mx_arr)
```

```python
import numpy as np

arrrr = np.array([1, 2, 3, 4, 5, -2, -14, -14, -34])

positive = list(filter(lambda x: x is not None,
                       map(lambda x: x if x > 0 else None, arrrr)))

negative = list(filter(lambda x: x is not None,
                       map(lambda x: x if x < 0 else None, arrrr)))

print("Positive:", positive)
print("Negative:", negative)
```
