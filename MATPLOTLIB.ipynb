```python
from matplotlib.animation import FuncAnimation

fig, ax = plt.subplots(figsize=(10, 5))
classes_index = np.arange(len(classes))
width = 0.2

# Initialize all bars with zero height. ax.bar returns a BarContainer, which is iterable over individual Rectangle patches.
bars1_patches = ax.bar(classes_index, [0]*len(classes), width, align='edge', color='m', edgecolor='m', linewidth=3, label='Style 1')
bars2_patches = ax.bar(classes_index + width, [0]*len(classes), width, align='edge', color='b', edgecolor='b', linewidth=3, label='Style 2')
bars3_patches = ax.bar(classes_index + width + width, [0]*len(classes), width, align='edge', color='r', edgecolor='r', linewidth=3, label='Style 3')

ax.set_xticks(classes_index + width / 2)
ax.set_xticklabels(classes, rotation=20)
ax.set_title('Animated Bar Chart of Styles per Class')
ax.set_xlabel('Classes')
ax.set_ylabel('Values')
ax.set_ylim(0, max(max(sty1), max(sty2), max(sty3)) * 1.1)

# Initialize legend once outside the update function.
# Use only one representative patch per bar group as handle for the legend.
legend_handles = [bars1_patches[0], bars2_patches[0], bars3_patches[0]]
current_legend = ax.legend(handles=legend_handles, labels=['', '', ''], loc='upper left')

def update(frame):
    artists_to_redraw = [] # This list will hold all artists modified in the current frame
    legend_texts = current_legend.get_texts() # Get the Text objects for the legend labels

    if frame == 0:
        for i, bar_patch in enumerate(bars1_patches):
            bar_patch.set_height(sty1[i]) # Set height of individual Rectangle patches
            artists_to_redraw.append(bar_patch)
        legend_texts[0].set_text('Style 1') # Update legend text using set_text()
        legend_texts[1].set_text('')
        legend_texts[2].set_text('')
        artists_to_redraw.extend(legend_texts)
    elif frame == 1:
        for i, bar_patch in enumerate(bars2_patches):
            bar_patch.set_height(sty2[i])
            artists_to_redraw.append(bar_patch)
        legend_texts[0].set_text('Style 1')
        legend_texts[1].set_text('Style 2')
        legend_texts[2].set_text('')
        artists_to_redraw.extend(legend_texts)
    elif frame == 2:
        for i, bar_patch in enumerate(bars3_patches):
            bar_patch.set_height(sty3[i])
            artists_to_redraw.append(bar_patch)
        legend_texts[0].set_text('Style 1')
        legend_texts[1].set_text('Style 2')
        legend_texts[2].set_text('Style 3')
        artists_to_redraw.extend(legend_texts)

    return tuple(artists_to_redraw) # FuncAnimation with blit=True expects a tuple of Artists

ani = FuncAnimation(fig, update, frames=range(3), blit=True, repeat=False, interval=1000)
plt.close(fig) # Prevent duplicate static plot

# To display the animation in Colab, we need to convert it to an HTML video.
# This might take a few seconds.
from IPython.display import HTML
HTML(ani.to_jshtml())
```

```python
import matplotlib.pyplot as plt

# Modern color palette
colors = ['#5B8FF9', '#61DDAA', '#65789B', '#F6BD16', '#7262FD']

plt.style.use('seaborn-v0_8-white')

fig, ax = plt.subplots(figsize=(5, 5), facecolor='white')

wedges, texts, autotexts = ax.pie(
    sty1,
    labels=classes,
    colors=colors,
    startangle=90,
    counterclock=False,
    autopct='%1.1f%%',
    pctdistance=0.75,
    labeldistance=1.08,
    explode=[0.02]*len(classes),
    wedgeprops=dict(
        width=0.55,              # Donut chart
        edgecolor='white',
        linewidth=3
    ),
    textprops=dict(
        fontsize=14,
        fontweight='semibold',
        color='#333333'
    )
)

# Style percentage labels
for autotext in autotexts:
    autotext.set_color("white")
    autotext.set_fontsize(13)
    autotext.set_fontweight("bold")

# Title
ax.set_title(
    "Distribution of Styles",
    fontsize=24,
    fontweight='bold',
    color='#222222',
    pad=25
)

# Center circle for cleaner donut appearance
centre_circle = plt.Circle((0, 0), 0.45, fc='white')
fig.gca().add_artist(centre_circle)

ax.axis('equal')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import random
```

```python
m1_stu_age = np.random.randint(16,30,(100))
py_stu_age = np.random.randint(14,28,(100))

m1_stu_age
```

```python
py_stu_age
```

```python
plt.hist(m1_stu_age , bins =5 , edgecolor = 'k')
plt.show()
```

```python
plt.hist(py_stu_age , bins =5 , edgecolor = 'k')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6), facecolor='white')

bins = np.arange(14, 31, 1)

ax.hist(m1_stu_age,
        bins=bins,
        color='#4E79A7',
        alpha=0.75,
        edgecolor='white',
        linewidth=1.5,
        label='Machine Learning')

ax.hist(py_stu_age,
        bins=bins,
        color='#F28E2B',
        alpha=0.75,
        edgecolor='white',
        linewidth=1.5,
        label='Python')

ax.set_title('Age Distribution of Students',
             fontsize=20,
             fontweight='bold',
             pad=15)

ax.set_xlabel('Age', fontsize=14)
ax.set_ylabel('Number of Students', fontsize=14)

ax.grid(axis='y', linestyle='--', alpha=0.4)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.legend(frameon=False, fontsize=12)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.hist([m1_stu_age, py_stu_age],
         bins=np.arange(14,31),
         color=['#2A9D8F','#E76F51'],
         edgecolor='white',
         linewidth=1.2,
         rwidth=0.9,
         label=['Machine Learning','Python'])

plt.title("Age Distribution", fontsize=20, weight='bold')
plt.xlabel("Age", fontsize=14)
plt.ylabel("Students", fontsize=14)

plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.legend(frameon=False)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

bins = np.arange(14,31)

plt.hist(m1_stu_age,
         bins=bins,
         histtype='step',
         linewidth=3,
         color='#0077B6',
         label='Machine Learning')

plt.hist(py_stu_age,
         bins=bins,
         histtype='step',
         linewidth=3,
         color='#E63946',
         label='Python')

plt.title("Age Distribution", fontsize=20, weight='bold')
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.grid(alpha=0.3)

plt.legend(frameon=False)

plt.tight_layout()
plt.show()
```

```python

fig, ax = plt.subplots(1, 2, figsize=(14, 6), facecolor='white')

# Vertical Histogram
ax[0].hist(
    m1_stu_age,
    bins=np.arange(16, 31),
    color='#4E79A7',
    edgecolor='white',
    linewidth=1.5
)
ax[0].set_title("Machine Learning Students", fontsize=16, fontweight='bold')
ax[0].set_xlabel("Age")
ax[0].set_ylabel("Frequency")
ax[0].grid(axis='y', linestyle='--', alpha=0.3)
ax[0].spines['top'].set_visible(False)
ax[0].spines['right'].set_visible(False)

# Horizontal Histogram
ax[1].hist(
    py_stu_age,
    bins=np.arange(14, 29),
    orientation='horizontal',
    color='#F28E2B',
    edgecolor='white',
    linewidth=1.5
)
ax[1].set_title("Python Students", fontsize=16, fontweight='bold')
ax[1].set_xlabel("Frequency")
ax[1].set_ylabel("Age")
ax[1].grid(axis='x', linestyle='--', alpha=0.3)
ax[1].spines['top'].set_visible(False)
ax[1].spines['right'].set_visible(False)

plt.suptitle("Age Distribution of Students", fontsize=20, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(5,5), facecolor='white')

bins = np.arange(14, 31)

# Vertical Histogram
plt.hist(
    m1_stu_age,
    bins=bins,
    color='#4E79A7',
    alpha=0.7,
    edgecolor='white',
    linewidth=1.5,
    label='Machine Learning'
)

# Horizontal Histogram
plt.hist(
    py_stu_age,
    bins=bins,
    orientation='horizontal',
    color='#F28E2B',
    alpha=0.7,
    edgecolor='white',
    linewidth=1.5,
    label='Python'
)

plt.title("Age Distribution", fontsize=20, fontweight='bold')
plt.xlabel("Frequency (Vertical Histogram)")
plt.ylabel("Age / Frequency (Horizontal Histogram)")

plt.grid(alpha=0.3, linestyle='--')
plt.legend(frameon=False)

plt.tight_layout()
plt.show()
```
